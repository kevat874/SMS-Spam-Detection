**Goal**: Load the original UCI SMS Spam Collection correctly and perform initial structural inspection.

Do **not** clean or model yet.

```python
import pandas as pd
import numpy as np
from pathlib import Path

RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)

PROJECT_ROOT = Path('..').resolve()
RAW_DATA_PATH = PROJECT_ROOT / 'data' / 'raw' / 'SMSSpamCollection'

print('Project root:', PROJECT_ROOT)
print('Raw data exists:', RAW_DATA_PATH.exists())
print('Raw data path:', RAW_DATA_PATH)
```

## Load the original dataset

- No header row
- Tab-separated ()
- Two columns: label + message
- We assign clean column names ourselves

```python
df = pd.read_csv(
    RAW_DATA_PATH,
    sep='\t',
    header=None,
    names=['label', 'message'],
    encoding='utf-8',
    quoting=3  # QUOTE_NONE – prevents issues with quotes inside messages
)

print('Shape:', df.shape)
print('\nColumns:', df.columns.tolist())
print('\nData types:')
print(df.dtypes)
print('\nFirst 10 rows:')
display(df.head(10))
```

```python
print('Total messages:', len(df))
print('\nLabel distribution:')
print(df['label'].value_counts())
print('\nLabel distribution (%):')
print(df['label'].value_counts(normalize=True).round(4) * 100)
print('\nMissing values:')
print(df.isnull().sum())
print('\nDuplicate rows:', df.duplicated().sum())
```

```python
print('Sample messages:')
for i, row in df.sample(5, random_state=RANDOM_SEED).iterrows():
    print(f'[{row["label"]}] {row["message"][:120]}...')
```
